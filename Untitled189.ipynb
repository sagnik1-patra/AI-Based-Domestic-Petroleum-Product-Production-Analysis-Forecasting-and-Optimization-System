```python
import json
import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT DIRECTORY
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Domestic Petroleum Product Production Analysis, Forecasting, and Optimization System"
)

DATA_PATH = BASE_DIR / "cleaned_petroleum_data.csv"
MODEL_PATH = BASE_DIR / "petroleum_model.pkl"
METADATA_PATH = BASE_DIR / "model_metadata.json"

BASE_DIR.mkdir(parents=True, exist_ok=True)

for path in [DATA_PATH, MODEL_PATH, METADATA_PATH]:
    if not path.exists():
        raise FileNotFoundError(
            f"Required file not found: {path}"
        )

# ============================================================
# 2. LOAD DATA AND METADATA
# ============================================================

df = pd.read_csv(DATA_PATH)

with open(METADATA_PATH, "r", encoding="utf-8") as f:
    metadata = json.load(f)

target = metadata["target_column"]
features = metadata["features"]

if target not in df.columns:
    raise ValueError(f"Target column missing: {target}")

missing_features = [
    col for col in features
    if col not in df.columns
]

if missing_features:
    raise ValueError(
        f"Missing feature columns: {missing_features}"
    )

df[target] = pd.to_numeric(
    df[target],
    errors="coerce"
)

df = df.dropna(subset=[target]).reset_index(drop=True)

print("Dataset Shape:", df.shape)
print("Target:", target)
print("Features:", features)

# ============================================================
# 3. RECONSTRUCT ORIGINAL TEST SPLIT
# ============================================================

from sklearn.model_selection import train_test_split

X = df[features].copy()
y = df[target].astype(float)

split_type = metadata.get("split_type", "random")

if split_type == "chronological_by_year":

    if "year" not in X.columns:
        raise ValueError(
            "Year column required for chronological split."
        )

    years = sorted(X["year"].dropna().unique())

    test_year = years[-1]

    test_mask = X["year"] == test_year

    X_test = X.loc[test_mask].copy()
    y_test = y.loc[test_mask].copy()

elif split_type == "chronological":

    order = pd.to_numeric(
        X["year"],
        errors="coerce"
    ).sort_values().index

    X = X.loc[order].reset_index(drop=True)
    y = y.loc[order].reset_index(drop=True)

    split_index = max(4, int(len(X) * 0.8))
    split_index = min(len(X) - 1, split_index)

    X_test = X.iloc[split_index:].copy()
    y_test = y.iloc[split_index:].copy()

else:

    _, X_test, _, y_test = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42
    )

print("\nTest Records:", len(X_test))

# IMPORTANT:
# This reconstructs the split from the previous training code.
# For exact reproducibility, saved test indices are preferable.
# If the original data order changed, rerun training.

# ============================================================
# 4. LOAD TRAINED MODEL
# ============================================================

model = joblib.load(MODEL_PATH)

print("\nLoaded:", MODEL_PATH.name)

predictions = model.predict(X_test)

actual = y_test.to_numpy(dtype=float)
predicted = np.asarray(predictions, dtype=float)

# ============================================================
# 5. EVALUATION METRICS
# ============================================================

def calculate_metrics(y_true, y_pred):

    mae = mean_absolute_error(y_true, y_pred)

    mse = mean_squared_error(y_true, y_pred)

    rmse = np.sqrt(mse)

    r2 = (
        r2_score(y_true, y_pred)
        if len(y_true) >= 2
        else np.nan
    )

    # MAPE only when actual values are nonzero
    nonzero = np.abs(y_true) > 1e-10

    if np.any(nonzero):
        mape = np.mean(
            np.abs(
                (y_true[nonzero] - y_pred[nonzero])
                / y_true[nonzero]
            )
        ) * 100
    else:
        mape = np.nan

    return {
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2": float(r2),
        "MAPE": float(mape)
    }


metrics = calculate_metrics(actual, predicted)

print("\nMODEL EVALUATION")
for name, value in metrics.items():
    print(f"{name}: {value:.4f}")

# ============================================================
# 6. LOAD MODEL COMPARISON
# ============================================================

comparison_path = BASE_DIR / "model_comparison.csv"

if comparison_path.exists():

    comparison = pd.read_csv(comparison_path)

else:

    comparison = pd.DataFrame([{
        "Model": metadata["best_ml_model"]["name"],
        "MAE": metrics["MAE"],
        "MSE": metrics["MSE"],
        "RMSE": metrics["RMSE"],
        "R2": metrics["R2"]
    }])

# ============================================================
# 7. SAVE RESULT CSV
# ============================================================

result_df = comparison.copy()

# Include neural network metrics if present in metadata.
nn_info = metadata.get("neural_network", {})

if nn_info and "RMSE" in nn_info:

    if not (
        result_df["Model"] == "Neural Network"
    ).any():

        nn_mae = nn_info.get("MAE", np.nan)
        nn_rmse = nn_info.get("RMSE", np.nan)
        nn_r2 = nn_info.get("R2", np.nan)

        result_df = pd.concat([
            result_df,
            pd.DataFrame([{
                "Model": "Neural Network",
                "MAE": nn_mae,
                "MSE": (
                    nn_rmse ** 2
                    if nn_rmse is not None
                    else np.nan
                ),
                "RMSE": nn_rmse,
                "R2": nn_r2
            }])
        ], ignore_index=True)

result_df.to_csv(
    BASE_DIR / "result.csv",
    index=False
)

print("\nSaved result.csv")
print(result_df.to_string(index=False))

# ============================================================
# 8. ACCURACY / PERFORMANCE GRAPH
# ============================================================

# R2 is a regression score, not classification accuracy.
# It can be negative and should not be clipped to 0-100%.

plt.figure(figsize=(10, 6))

valid = result_df.dropna(subset=["R2"])

bars = plt.bar(
    valid["Model"],
    valid["R2"],
    color="steelblue",
    edgecolor="black"
)

plt.axhline(
    y=0,
    color="black",
    linewidth=1
)

for bar, score in zip(bars, valid["R2"]):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{score:.3f}",
        ha="center",
        va="bottom" if score >= 0 else "top"
    )

plt.title(
    "Petroleum Production Model Performance (R²)",
    fontsize=14
)

plt.xlabel("Machine Learning Models")
plt.ylabel("R² Score")

plt.xticks(rotation=20)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved accuracy_graph.png")

# ============================================================
# 9. CORRELATION HEATMAP
# ============================================================

numeric_df = df.select_dtypes(
    include=np.number
).copy()

# Avoid plotting huge matrices
if numeric_df.shape[1] > 25:

    correlations_with_target = (
        numeric_df.corr()[target]
        .abs()
        .sort_values(ascending=False)
    )

    selected_columns = (
        correlations_with_target
        .head(25)
        .index
        .tolist()
    )

    numeric_df = numeric_df[selected_columns]

correlation_matrix = numeric_df.corr()

if correlation_matrix.empty:
    print("No numeric correlations available.")

else:

    plt.figure(figsize=(12, 9))

    plt.imshow(
        correlation_matrix,
        cmap="coolwarm",
        vmin=-1,
        vmax=1,
        aspect="auto"
    )

    plt.colorbar(label="Correlation Coefficient")

    plt.xticks(
        range(len(correlation_matrix.columns)),
        correlation_matrix.columns,
        rotation=90,
        fontsize=8
    )

    plt.yticks(
        range(len(correlation_matrix.index)),
        correlation_matrix.index,
        fontsize=8
    )

    if len(correlation_matrix) <= 15:

        for i in range(len(correlation_matrix)):
            for j in range(len(correlation_matrix)):

                value = correlation_matrix.iloc[i, j]

                if pd.notna(value):

                    plt.text(
                        j, i,
                        f"{value:.2f}",
                        ha="center",
                        va="center",
                        fontsize=7
                    )

    plt.title(
        "Correlation Heatmap of Petroleum Production Data"
    )

    plt.tight_layout()

    plt.savefig(
        BASE_DIR / "heatmap.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    correlation_matrix.to_csv(
        BASE_DIR / "correlation_matrix.csv"
    )

    print("Saved heatmap.png")

# ============================================================
# 10. MODEL COMPARISON GRAPH
# ============================================================

plot_data = result_df.dropna(
    subset=["MAE", "RMSE"]
)

x = np.arange(len(plot_data))
width = 0.35

plt.figure(figsize=(11, 6))

plt.bar(
    x - width / 2,
    plot_data["MAE"],
    width,
    label="MAE",
    color="royalblue"
)

plt.bar(
    x + width / 2,
    plot_data["RMSE"],
    width,
    label="RMSE",
    color="darkorange"
)

plt.xticks(
    x,
    plot_data["Model"],
    rotation=20
)

plt.xlabel("Models")
plt.ylabel("Prediction Error")
plt.title("Model Comparison: MAE and RMSE")

plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved comparison_graph.png")

# ============================================================
# 11. SAVE PREDICTION CSV
# ============================================================

prediction_df = X_test.copy().reset_index(drop=True)

prediction_df["actual_production"] = actual
prediction_df["predicted_production"] = predicted

prediction_df["absolute_error"] = np.abs(
    actual - predicted
)

prediction_df["percentage_error"] = np.where(
    np.abs(actual) > 1e-10,
    np.abs(actual - predicted)
    / np.maximum(np.abs(actual), 1e-10) * 100,
    np.nan
)

prediction_df.to_csv(
    BASE_DIR / "prediction.csv",
    index=False
)

print("\nSaved prediction.csv")
print(prediction_df.head(10).to_string(index=False))

# ============================================================
# 12. PREDICTION GRAPH
# ============================================================

# Sort by year if available; otherwise use test row order.

plot_df = prediction_df.copy()

if "year" in plot_df.columns:
    plot_df = plot_df.sort_values("year")

plt.figure(figsize=(13, 6))

plt.plot(
    range(len(plot_df)),
    plot_df["actual_production"],
    marker="o",
    linewidth=2,
    label="Actual Production",
    color="royalblue"
)

plt.plot(
    range(len(plot_df)),
    plot_df["predicted_production"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="Predicted Production",
    color="darkorange"
)

plt.xlabel("Test Observation")
plt.ylabel("Production")
plt.title("Actual vs Predicted Petroleum Production")

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved prediction_graph.png")

# ============================================================
# 13. RESULT GRAPH: ACTUAL VS PREDICTED SCATTER
# ============================================================

plt.figure(figsize=(9, 7))

plt.scatter(
    actual,
    predicted,
    alpha=0.75,
    color="teal",
    edgecolor="black",
    label="Predictions"
)

min_value = min(actual.min(), predicted.min())
max_value = max(actual.max(), predicted.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    "r--",
    linewidth=2,
    label="Perfect Prediction"
)

plt.xlabel("Actual Production")
plt.ylabel("Predicted Production")

plt.title(
    "Petroleum Production Prediction Results"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved result_graph.png")

# ============================================================
# 14. ADDITIONAL GRAPH: RESIDUAL ANALYSIS
# ============================================================

residuals = actual - predicted

plt.figure(figsize=(10, 6))

plt.scatter(
    predicted,
    residuals,
    color="purple",
    alpha=0.7
)

plt.axhline(
    y=0,
    color="red",
    linestyle="--"
)

plt.xlabel("Predicted Production")
plt.ylabel("Residual (Actual - Predicted)")
plt.title("Residual Analysis")

plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "residual_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved residual_graph.png")

# ============================================================
# 15. ADDITIONAL GRAPH: TRAINING HISTORY
# ============================================================

history_path = BASE_DIR / "training_history.csv"

if history_path.exists():

    history_df = pd.read_csv(history_path)

    plt.figure(figsize=(10, 6))

    if "loss" in history_df.columns:

        plt.plot(
            history_df["loss"],
            label="Training Loss",
            linewidth=2
        )

    if "val_loss" in history_df.columns:

        plt.plot(
            history_df["val_loss"],
            label="Validation Loss",
            linewidth=2
        )

    plt.xlabel("Epoch")
    plt.ylabel("Loss (Scaled MSE)")
    plt.title("Neural Network Training History")

    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()

    plt.savefig(
        BASE_DIR / "training_graph.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    print("Saved training_graph.png")

# ============================================================
# 16. SAVE EVALUATION METADATA
# ============================================================

evaluation_metadata = {
    "model": metadata["best_ml_model"]["name"],
    "target_column": target,
    "test_records": int(len(y_test)),
    "split_type": split_type,
    "metrics": {
        name: (
            None if pd.isna(value) else float(value)
        )
        for name, value in metrics.items()
    },
    "generated_files": [
        "accuracy_graph.png",
        "heatmap.png",
        "comparison_graph.png",
        "result.csv",
        "result_graph.png",
        "prediction.csv",
        "prediction_graph.png",
        "residual_graph.png"
    ]
}

with open(
    BASE_DIR / "evaluation_metadata.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        evaluation_metadata,
        f,
        indent=4
    )

# ============================================================
# 17. VERIFY ALL OUTPUTS
# ============================================================

print("\n" + "=" * 65)
print("PETROLEUM PRODUCTION ANALYSIS COMPLETED")
print("=" * 65)

output_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "residual_graph.png",
    "evaluation_metadata.json"
]

for filename in output_files:

    path = BASE_DIR / filename

    if path.exists():
        print(f"[SUCCESS] {filename}")
    else:
        print(f"[MISSING] {filename}")

print("\nFinal Evaluation Metrics:")

for key, value in metrics.items():
    print(f"{key}: {value:.4f}")

print("\nAll outputs saved in:")
print(BASE_DIR)
```
