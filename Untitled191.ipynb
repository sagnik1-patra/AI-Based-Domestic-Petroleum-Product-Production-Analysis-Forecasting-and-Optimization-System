```python

# ============================================================
# PSO-BASED DOMESTIC PETROLEUM PRODUCTION FORECASTING
# FIXED FOR ACTUAL EXCEL FILE: 2000-01 TO 2010-11(P)
# ALL OUTPUTS USE pso_ PREFIX
# ============================================================

import re
import json
import yaml
import joblib
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from datetime import datetime
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. SETTINGS
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Domestic Petroleum Product Production Analysis, Forecasting, and Optimization System"
)

DATA_PATH = (
    BASE_DIR /
    "DOMESTIC_PRODUCTION_OF_PETROLEUM_PRODUCTS_1.xls"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42
rng = np.random.default_rng(SEED)

PARTICLES = 8
ITERATIONS = 12

INERTIA = 0.7
COGNITIVE = 1.5
SOCIAL = 1.5

SHOW_GRAPHS = True

if not DATA_PATH.exists():
    raise FileNotFoundError(DATA_PATH)

# ============================================================
# 2. LOAD ORIGINAL EXCEL DATA
# ============================================================

df = pd.read_excel(
    DATA_PATH,
    engine="xlrd"
)

df.columns = [
    re.sub(
        r"[^a-z0-9]+",
        "_",
        str(c).strip().lower()
    ).strip("_")
    for c in df.columns
]

print("Original Columns:")
print(df.columns.tolist())

if "year" not in df.columns:
    raise ValueError("Year column not found.")

if "total" not in df.columns:
    raise ValueError("Total column not found.")

# Financial-year examples:
# 2000-01 -> 2000
# 2010-11(P) -> 2010

def extract_year(value):
    match = re.search(
        r"(19\d{2}|20\d{2})",
        str(value)
    )
    return (
        int(match.group(1))
        if match else np.nan
    )

df["financial_year"] = df["year"].astype(str)

df["year"] = df["year"].apply(
    extract_year
)

df["total"] = pd.to_numeric(
    df["total"],
    errors="coerce"
)

df = df.dropna(
    subset=["year", "total"]
).copy()

df["year"] = df["year"].astype(int)

df = df.sort_values(
    "year"
).reset_index(drop=True)

if len(df) < 8:
    raise ValueError(
        "At least 8 valid annual observations "
        "are required for this PSO experiment."
    )

if df["year"].duplicated().any():
    raise ValueError(
        "Duplicate years found. Aggregate or "
        "resolve duplicate records first."
    )

print("\nCleaned Data:")
print(
    df[
        ["financial_year", "year", "total"]
    ].to_string(index=False)
)

df.to_csv(
    BASE_DIR / "pso_cleaned_data.csv",
    index=False
)

# ============================================================
# 3. DEFINE FORECASTING FEATURES
# ============================================================

# Only the historical year is used as input.
# Petroleum components and annual growth are excluded
# to avoid leaking information about the target.

FEATURES = ["year"]
TARGET = "total"

X = df[FEATURES].copy()
y = df[TARGET].astype(float)

# Last two financial years are held out for testing.
TEST_SIZE = 2

X_train = X.iloc[:-TEST_SIZE].copy()
X_test = X.iloc[-TEST_SIZE:].copy()

y_train = y.iloc[:-TEST_SIZE].copy()
y_test = y.iloc[-TEST_SIZE:].copy()

print("\nTraining Records:", len(X_train))
print("Testing Records:", len(X_test))

print(
    "Test Years:",
    X_test["year"].tolist()
)

# ============================================================
# 4. EXPANDING-WINDOW VALIDATION
# ============================================================

# Train on earlier years, validate on the next year.
# Validation uses training data only.
# The final two test years remain unseen.

MIN_TRAIN_YEARS = 5

validation_folds = []

for end in range(
    MIN_TRAIN_YEARS,
    len(X_train)
):

    train_indices = np.arange(end)
    val_indices = np.array([end])

    validation_folds.append(
        (train_indices, val_indices)
    )

if not validation_folds:
    raise ValueError(
        "Insufficient observations for validation."
    )

print(
    "\nExpanding-Window Validation Folds:",
    len(validation_folds)
)

# ============================================================
# 5. PSO HYPERPARAMETER SEARCH SPACE
# ============================================================

# Particle dimensions:
# 0: n_estimators
# 1: max_depth
# 2: min_samples_split
# 3: min_samples_leaf

LOWER = np.array(
    [30, 2, 2, 1],
    dtype=float
)

UPPER = np.array(
    [200, 12, 6, 3],
    dtype=float
)

DIMENSIONS = len(LOWER)

def decode_particle(position):

    p = np.clip(
        position,
        LOWER,
        UPPER
    )

    return {
        "n_estimators": int(round(p[0])),
        "max_depth": int(round(p[1])),
        "min_samples_split": int(round(p[2])),
        "min_samples_leaf": int(round(p[3])),
        "random_state": SEED,
        "n_jobs": -1
    }

# ============================================================
# 6. FITNESS FUNCTION
# ============================================================

fitness_cache = {}

def fitness(position):

    params = decode_particle(position)

    key = tuple(
        params[k]
        for k in [
            "n_estimators",
            "max_depth",
            "min_samples_split",
            "min_samples_leaf"
        ]
    )

    if key in fitness_cache:
        return fitness_cache[key]

    actual_values = []
    predicted_values = []

    for train_idx, val_idx in validation_folds:

        model = RandomForestRegressor(
            **params
        )

        model.fit(
            X_train.iloc[train_idx],
            y_train.iloc[train_idx]
        )

        pred = model.predict(
            X_train.iloc[val_idx]
        )

        actual_values.extend(
            y_train.iloc[val_idx].tolist()
        )

        predicted_values.extend(
            pred.tolist()
        )

    rmse = np.sqrt(
        mean_squared_error(
            actual_values,
            predicted_values
        )
    )

    fitness_cache[key] = float(rmse)

    return float(rmse)

# ============================================================
# 7. INITIALIZE PSO
# ============================================================

positions = rng.uniform(
    LOWER,
    UPPER,
    size=(PARTICLES, DIMENSIONS)
)

velocity_limit = (
    UPPER - LOWER
) * 0.20

velocities = rng.uniform(
    -velocity_limit,
    velocity_limit,
    size=(PARTICLES, DIMENSIONS)
)

personal_best = positions.copy()

personal_scores = np.array([
    fitness(p)
    for p in positions
])

best_index = np.argmin(
    personal_scores
)

global_best = personal_best[
    best_index
].copy()

global_score = float(
    personal_scores[best_index]
)

history = []

# ============================================================
# 8. RUN PSO OPTIMIZATION
# ============================================================

print("\nStarting PSO Optimization...")

for iteration in range(ITERATIONS):

    for i in range(PARTICLES):

        r1 = rng.random(DIMENSIONS)
        r2 = rng.random(DIMENSIONS)

        velocities[i] = (
            INERTIA * velocities[i]
            +
            COGNITIVE * r1 *
            (personal_best[i] - positions[i])
            +
            SOCIAL * r2 *
            (global_best - positions[i])
        )

        velocities[i] = np.clip(
            velocities[i],
            -velocity_limit,
            velocity_limit
        )

        positions[i] += velocities[i]

        positions[i] = np.clip(
            positions[i],
            LOWER,
            UPPER
        )

        score = fitness(
            positions[i]
        )

        if score < personal_scores[i]:

            personal_scores[i] = score
            personal_best[i] = positions[i].copy()

            if score < global_score:
                global_score = score
                global_best = positions[i].copy()

    history.append({
        "Iteration": iteration + 1,
        "Best_RMSE": global_score
    })

    print(
        f"Iteration {iteration + 1}/{ITERATIONS} "
        f"| Validation RMSE: {global_score:.5f}"
    )

best_params = decode_particle(
    global_best
)

print("\nBest Parameters:")
print(best_params)

# ============================================================
# 9. SAVE CONVERGENCE GRAPH
# ============================================================

history_df = pd.DataFrame(history)

history_df.to_csv(
    BASE_DIR / "pso_optimization_history.csv",
    index=False
)

def save_show(filename):
    plt.tight_layout()
    plt.savefig(
        BASE_DIR / filename,
        dpi=300,
        bbox_inches="tight"
    )
    if SHOW_GRAPHS:
        plt.show()
    else:
        plt.close()

plt.figure(figsize=(10, 6))

plt.plot(
    history_df["Iteration"],
    history_df["Best_RMSE"],
    marker="o",
    linewidth=2
)

plt.title("PSO Optimization Convergence")
plt.xlabel("Iteration")
plt.ylabel("Validation RMSE")
plt.grid(alpha=0.3)

save_show("pso_convergence_graph.png")

# ============================================================
# 10. TRAIN BASELINE AND PSO MODELS
# ============================================================

baseline_model = RandomForestRegressor(
    n_estimators=100,
    random_state=SEED,
    n_jobs=-1
)

baseline_model.fit(
    X_train,
    y_train
)

pso_model = RandomForestRegressor(
    **best_params
)

pso_model.fit(
    X_train,
    y_train
)

baseline_pred = baseline_model.predict(
    X_test
)

pso_pred = pso_model.predict(
    X_test
)

joblib.dump(
    pso_model,
    BASE_DIR / "pso_petroleum_model.pkl"
)

# ============================================================
# 11. EVALUATION METRICS
# ============================================================

def evaluate(actual, predicted):

    actual = np.asarray(
        actual,
        dtype=float
    )

    predicted = np.asarray(
        predicted,
        dtype=float
    )

    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(mse)

    r2 = (
        r2_score(actual, predicted)
        if len(actual) >= 2
        else np.nan
    )

    valid = np.abs(actual) > 1e-10

    mape = (
        np.mean(
            np.abs(
                (actual[valid] - predicted[valid])
                / actual[valid]
            )
        ) * 100
        if valid.any()
        else np.nan
    )

    return {
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2": float(r2),
        "MAPE": float(mape)
    }

baseline_metrics = evaluate(
    y_test,
    baseline_pred
)

pso_metrics = evaluate(
    y_test,
    pso_pred
)

result_df = pd.DataFrame([
    {
        "Model": "Baseline Random Forest",
        **baseline_metrics
    },
    {
        "Model": "PSO Optimized Random Forest",
        **pso_metrics
    }
])

result_df.to_csv(
    BASE_DIR / "pso_result.csv",
    index=False
)

print("\nModel Evaluation:")
print(result_df.to_string(index=False))

# ============================================================
# 12. SAVE PREDICTION CSV
# ============================================================

prediction_df = pd.DataFrame({
    "financial_year": df["financial_year"].iloc[
        -TEST_SIZE:
    ].to_numpy(),
    "year": X_test["year"].to_numpy(),
    "actual_production": y_test.to_numpy(),
    "baseline_prediction": baseline_pred,
    "predicted_production": pso_pred
})

prediction_df["absolute_error"] = np.abs(
    prediction_df["actual_production"] -
    prediction_df["predicted_production"]
)

prediction_df["percentage_error"] = (
    prediction_df["absolute_error"] /
    prediction_df["actual_production"].abs()
    .replace(0, np.nan)
) * 100

prediction_df.to_csv(
    BASE_DIR / "pso_prediction.csv",
    index=False
)

# ============================================================
# 13. ACCURACY / R2 GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

bars = plt.bar(
    result_df["Model"],
    result_df["R2"],
    color=["steelblue", "seagreen"]
)

plt.axhline(0, color="black")

for bar, score in zip(
    bars,
    result_df["R2"]
):
    if np.isfinite(score):
        plt.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height(),
            f"{score:.3f}",
            ha="center",
            va="bottom" if score >= 0 else "top"
        )

plt.title("PSO Model Performance (R²)")
plt.ylabel("R² Score")
plt.xticks(rotation=10)
plt.grid(axis="y", alpha=0.3)

save_show("pso_accuracy_graph.png")

# ============================================================
# 14. CORRELATION HEATMAP
# ============================================================

numeric_df = df.select_dtypes(
    include=np.number
)

corr = numeric_df.corr()

corr.to_csv(
    BASE_DIR / "pso_correlation_matrix.csv"
)

plt.figure(figsize=(13, 10))

plt.imshow(
    corr,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    aspect="auto"
)

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=90,
    fontsize=8
)

plt.yticks(
    range(len(corr.index)),
    corr.index,
    fontsize=8
)

if len(corr) <= 18:
    for i in range(len(corr)):
        for j in range(len(corr)):
            value = corr.iloc[i, j]
            if pd.notna(value):
                plt.text(
                    j, i,
                    f"{value:.2f}",
                    ha="center",
                    va="center",
                    fontsize=6
                )

plt.title("Petroleum Production Correlation Heatmap")

save_show("pso_heatmap.png")

# ============================================================
# 15. MODEL COMPARISON GRAPH
# ============================================================

x = np.arange(len(result_df))
width = 0.35

plt.figure(figsize=(11, 6))

plt.bar(
    x - width / 2,
    result_df["MAE"],
    width,
    label="MAE",
    color="royalblue"
)

plt.bar(
    x + width / 2,
    result_df["RMSE"],
    width,
    label="RMSE",
    color="darkorange"
)

plt.xticks(
    x,
    result_df["Model"],
    rotation=10
)

plt.title("Baseline vs PSO Model Comparison")
plt.ylabel("Prediction Error")
plt.legend()
plt.grid(axis="y", alpha=0.3)

save_show("pso_comparison_graph.png")

# ============================================================
# 16. PREDICTION GRAPH
# ============================================================

plt.figure(figsize=(11, 6))

plt.plot(
    prediction_df["financial_year"],
    prediction_df["actual_production"],
    marker="o",
    label="Actual Production",
    linewidth=2
)

plt.plot(
    prediction_df["financial_year"],
    prediction_df["predicted_production"],
    marker="s",
    linestyle="--",
    label="PSO Prediction",
    linewidth=2
)

plt.title("PSO Actual vs Predicted Production")
plt.xlabel("Financial Year")
plt.ylabel("Total Production")
plt.legend()
plt.grid(alpha=0.3)

save_show("pso_prediction_graph.png")

# ============================================================
# 17. RESULT GRAPH
# ============================================================

actual = y_test.to_numpy(dtype=float)
predicted = np.asarray(pso_pred, dtype=float)

plt.figure(figsize=(9, 7))

plt.scatter(
    actual,
    predicted,
    color="seagreen",
    s=100,
    edgecolor="black"
)

low = min(actual.min(), predicted.min())
high = max(actual.max(), predicted.max())

plt.plot(
    [low, high],
    [low, high],
    "r--",
    label="Perfect Prediction"
)

plt.title("PSO Prediction Results")
plt.xlabel("Actual Production")
plt.ylabel("Predicted Production")
plt.legend()
plt.grid(alpha=0.3)

save_show("pso_result_graph.png")

# ============================================================
# 18. RESIDUAL GRAPH
# ============================================================

residuals = actual - predicted

plt.figure(figsize=(10, 6))

plt.scatter(
    predicted,
    residuals,
    color="purple",
    s=90
)

plt.axhline(
    0,
    color="red",
    linestyle="--"
)

plt.title("PSO Residual Analysis")
plt.xlabel("Predicted Production")
plt.ylabel("Residual")
plt.grid(alpha=0.3)

save_show("pso_residual_graph.png")

# ============================================================
# 19. HISTORICAL PRODUCTION GRAPH
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    df["year"],
    df["total"],
    marker="o",
    linewidth=2,
    color="navy"
)

plt.title("Historical Domestic Petroleum Production")
plt.xlabel("Financial Year Starting Year")
plt.ylabel("Total Production")
plt.grid(alpha=0.3)

save_show("pso_historical_graph.png")

# ============================================================
# 20. FUTURE-YEAR PREDICTIONS
# ============================================================

# Random Forest cannot extrapolate trends reliably.
# These values are exploratory, not validated forecasts.

last_year = int(df["year"].max())

future_years = np.arange(
    last_year + 1,
    last_year + 4
)

future_X = pd.DataFrame({
    "year": future_years
})

# Refit using all available historical data
# for exploratory future predictions.

future_model = RandomForestRegressor(
    **best_params
)

future_model.fit(X, y)

future_predictions = future_model.predict(
    future_X
)

future_df = pd.DataFrame({
    "year": future_years,
    "predicted_production": future_predictions
})

future_df.to_csv(
    BASE_DIR / "pso_future_prediction.csv",
    index=False
)

joblib.dump(
    future_model,
    BASE_DIR / "pso_future_model.pkl"
)

plt.figure(figsize=(12, 6))

plt.plot(
    df["year"],
    df["total"],
    marker="o",
    label="Historical Production"
)

plt.plot(
    future_years,
    future_predictions,
    marker="s",
    linestyle="--",
    label="Exploratory PSO Projection"
)

plt.title("Historical Production and PSO Projection")
plt.xlabel("Year")
plt.ylabel("Total Production")
plt.legend()
plt.grid(alpha=0.3)

save_show("pso_future_prediction_graph.png")

# ============================================================
# 21. SAVE YAML CONFIG
# ============================================================

config = {
    "project": (
        "PSO-Based Domestic Petroleum "
        "Production Forecasting"
    ),
    "algorithm": "Particle Swarm Optimization",
    "model": "RandomForestRegressor",
    "dataset": DATA_PATH.name,
    "target": TARGET,
    "features": FEATURES,
    "training_records": len(X_train),
    "testing_records": len(X_test),
    "split": "last_two_financial_years",
    "validation": "expanding_window",
    "pso": {
        "particles": PARTICLES,
        "iterations": ITERATIONS,
        "inertia": INERTIA,
        "cognitive": COGNITIVE,
        "social": SOCIAL,
        "random_state": SEED
    },
    "best_parameters": best_params,
    "best_validation_rmse": float(global_score),
    "test_metrics": pso_metrics
}

with open(
    BASE_DIR / "pso_config.yaml",
    "w",
    encoding="utf-8"
) as f:
    yaml.safe_dump(
        config,
        f,
        sort_keys=False
    )

# ============================================================
# 22. SAVE JSON METADATA
# ============================================================

metadata = {
    "project": config["project"],
    "created_at": datetime.now().isoformat(),
    "algorithm": "PSO",
    "base_model": "RandomForestRegressor",
    "target": TARGET,
    "features": FEATURES,
    "total_records": len(df),
    "training_records": len(X_train),
    "testing_records": len(X_test),
    "validation_folds": len(validation_folds),
    "best_parameters": best_params,
    "validation_rmse": float(global_score),
    "test_metrics": pso_metrics,
    "baseline_metrics": baseline_metrics,
    "note": (
        "Small historical dataset. "
        "Future Random Forest projections "
        "are exploratory and cannot be "
        "treated as validated extrapolations."
    )
}

with open(
    BASE_DIR / "pso_model_metadata.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metadata,
        f,
        indent=4,
        allow_nan=False
    )

# ============================================================
# 23. VERIFY ALL GENERATED FILES
# ============================================================

output_files = [
    "pso_petroleum_model.pkl",
    "pso_future_model.pkl",
    "pso_config.yaml",
    "pso_model_metadata.json",
    "pso_cleaned_data.csv",
    "pso_accuracy_graph.png",
    "pso_heatmap.png",
    "pso_comparison_graph.png",
    "pso_result.csv",
    "pso_result_graph.png",
    "pso_prediction.csv",
    "pso_prediction_graph.png",
    "pso_convergence_graph.png",
    "pso_residual_graph.png",
    "pso_historical_graph.png",
    "pso_future_prediction.csv",
    "pso_future_prediction_graph.png",
    "pso_optimization_history.csv",
    "pso_correlation_matrix.csv"
]

print("\n" + "=" * 65)
print("PSO PROJECT COMPLETED")
print("=" * 65)

for filename in output_files:

    path = BASE_DIR / filename

    if path.exists():
        print("[SUCCESS]", filename)
    else:
        print("[MISSING]", filename)

print("\nBest Parameters:")
print(best_params)

print("\nValidation RMSE:", round(global_score, 5))

print("\nTest Metrics:")
for key, value in pso_metrics.items():
    print(f"{key}: {value:.5f}")

print("\nOutput Directory:")
print(BASE_DIR)
```
