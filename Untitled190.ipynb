```python
import json
import yaml
import joblib
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from datetime import datetime

from sklearn.model_selection import (
    train_test_split,
    KFold,
    TimeSeriesSplit,
    cross_val_score
)
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT PATHS
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Domestic Petroleum Product Production Analysis, Forecasting, and Optimization System"
)

DATA_PATH = BASE_DIR / "cleaned_petroleum_data.csv"
METADATA_PATH = BASE_DIR / "model_metadata.json"

BASE_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42
rng = np.random.default_rng(SEED)

# AIS settings
POPULATION_SIZE = 10
GENERATIONS = 12
SELECTED_ANTIBODIES = 4
CLONES_PER_ANTIBODY = 3
RANDOM_REPLACEMENTS = 2

# Use more generations for larger experiments.
# Small values make initial execution faster.

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Missing dataset: {DATA_PATH}"
    )

if not METADATA_PATH.exists():
    raise FileNotFoundError(
        f"Missing metadata: {METADATA_PATH}"
    )

print("=" * 65)
print("AIS-BASED PETROLEUM PRODUCTION OPTIMIZATION")
print("=" * 65)

# ============================================================
# 2. LOAD DATA
# ============================================================

df = pd.read_csv(DATA_PATH)

with open(METADATA_PATH, "r", encoding="utf-8") as f:
    metadata = json.load(f)

TARGET = metadata["target_column"]
FEATURES = metadata["features"]

missing = [
    col for col in FEATURES + [TARGET]
    if col not in df.columns
]

if missing:
    raise ValueError(
        f"Missing columns: {missing}"
    )

df[TARGET] = pd.to_numeric(
    df[TARGET],
    errors="coerce"
)

df = df.dropna(
    subset=[TARGET]
).reset_index(drop=True)

X = df[FEATURES].copy()
y = df[TARGET].astype(float)

if len(df) < 10:
    raise ValueError(
        "At least 10 valid observations are required "
        "for this AIS demonstration. Do not duplicate "
        "observations to increase sample size."
    )

print("\nDataset Shape:", df.shape)
print("Target:", TARGET)
print("Features:", FEATURES)

# ============================================================
# 3. TRAIN-TEST SPLIT
# ============================================================

split_type = metadata.get(
    "split_type",
    "random"
)

if split_type == "chronological_by_year":

    if "year" not in X.columns:
        raise ValueError(
            "Missing year column for chronological split."
        )

    years = sorted(X["year"].dropna().unique())
    test_year = years[-1]

    train_mask = X["year"] < test_year
    test_mask = X["year"] == test_year

    X_train = X.loc[train_mask].copy()
    X_test = X.loc[test_mask].copy()

    y_train = y.loc[train_mask].copy()
    y_test = y.loc[test_mask].copy()

    temporal = True

elif split_type == "chronological":

    if "year" not in X.columns:
        raise ValueError(
            "Missing year column."
        )

    order = pd.to_numeric(
        X["year"],
        errors="coerce"
    ).sort_values().index

    X = X.loc[order].reset_index(drop=True)
    y = y.loc[order].reset_index(drop=True)

    split_index = max(4, int(len(X) * 0.8))
    split_index = min(len(X) - 1, split_index)

    X_train = X.iloc[:split_index].copy()
    X_test = X.iloc[split_index:].copy()

    y_train = y.iloc[:split_index].copy()
    y_test = y.iloc[split_index:].copy()

    temporal = True

else:

    X_train, X_test, y_train, y_test = (
        train_test_split(
            X,
            y,
            test_size=0.20,
            random_state=SEED
        )
    )

    temporal = False

if len(X_train) < 6 or len(X_test) < 1:
    raise ValueError(
        "Insufficient train/test observations."
    )

print("\nTraining Records:", len(X_train))
print("Testing Records:", len(X_test))
print("Split Type:", split_type)

# ============================================================
# 4. PREPROCESSING
# ============================================================

numeric_features = X.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = [
    col for col in X.columns
    if col not in numeric_features
]

def build_preprocessor():

    transformers = []

    if numeric_features:

        numeric_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "scaler",
                StandardScaler()
            )
        ])

        transformers.append((
            "numeric",
            numeric_pipeline,
            numeric_features
        ))

    if categorical_features:

        categorical_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(
                    strategy="constant",
                    fill_value="Unknown"
                )
            ),
            (
                "encoder",
                OneHotEncoder(
                    handle_unknown="ignore",
                    sparse_output=False
                )
            )
        ])

        transformers.append((
            "categorical",
            categorical_pipeline,
            categorical_features
        ))

    return ColumnTransformer(
        transformers=transformers,
        remainder="drop"
    )

# ============================================================
# 5. AIS SEARCH SPACE
# ============================================================

# Each antibody represents:
# [n_estimators, max_depth, min_samples_split,
#  min_samples_leaf, max_features_fraction]

LOWER_BOUNDS = np.array([
    50, 3, 2, 1, 0.3
], dtype=float)

UPPER_BOUNDS = np.array([
    300, 25, 15, 8, 1.0
], dtype=float)

def random_antibody():

    return rng.uniform(
        LOWER_BOUNDS,
        UPPER_BOUNDS
    )


def decode_antibody(antibody):

    a = np.clip(
        antibody,
        LOWER_BOUNDS,
        UPPER_BOUNDS
    )

    return {
        "n_estimators": int(round(a[0])),
        "max_depth": int(round(a[1])),
        "min_samples_split": int(round(a[2])),
        "min_samples_leaf": int(round(a[3])),
        "max_features": float(a[4]),
        "random_state": SEED,
        "n_jobs": -1
    }


def build_model(params):

    return Pipeline([
        (
            "preprocessor",
            build_preprocessor()
        ),
        (
            "model",
            RandomForestRegressor(**params)
        )
    ])

# ============================================================
# 6. VALIDATION STRATEGY
# ============================================================

if temporal:

    # Preserve chronological order.
    # For panel data with multiple rows per year,
    # year-grouped validation is preferable.

    n_splits = min(3, len(X_train) - 2)

    if n_splits < 2:
        raise ValueError(
            "Insufficient training records for "
            "temporal cross-validation."
        )

    cv = TimeSeriesSplit(
        n_splits=n_splits
    )

else:

    n_splits = min(3, len(X_train))

    cv = KFold(
        n_splits=n_splits,
        shuffle=True,
        random_state=SEED
    )

# ============================================================
# 7. FITNESS FUNCTION
# ============================================================

fitness_cache = {}

def fitness(antibody):

    params = decode_antibody(antibody)

    key = (
        params["n_estimators"],
        params["max_depth"],
        params["min_samples_split"],
        params["min_samples_leaf"],
        round(params["max_features"], 4)
    )

    if key in fitness_cache:
        return fitness_cache[key]

    model = build_model(params)

    try:

        scores = cross_val_score(
            model,
            X_train,
            y_train,
            scoring="neg_mean_squared_error",
            cv=cv,
            n_jobs=1,
            error_score="raise"
        )

        # RMSE aggregated across validation folds
        rmse = float(
            np.sqrt(np.mean(-scores))
        )

    except Exception as exc:

        print("Candidate failed:", str(exc)[:120])
        rmse = float("inf")

    fitness_cache[key] = rmse

    return rmse

# ============================================================
# 8. HYPERMUTATION
# ============================================================

def hypermutate(antibody, mutation_strength):

    scale = (
        UPPER_BOUNDS - LOWER_BOUNDS
    )

    mutation = rng.normal(
        loc=0,
        scale=mutation_strength,
        size=len(antibody)
    )

    mutated = antibody + mutation * scale

    return np.clip(
        mutated,
        LOWER_BOUNDS,
        UPPER_BOUNDS
    )

# ============================================================
# 9. ARTIFICIAL IMMUNE SYSTEM OPTIMIZATION
# ============================================================

population = [
    random_antibody()
    for _ in range(POPULATION_SIZE)
]

history = []

global_best = None
global_best_score = float("inf")

print("\nStarting AIS Optimization...")

for generation in range(GENERATIONS):

    evaluated = [
        (fitness(a), a.copy())
        for a in population
    ]

    evaluated.sort(
        key=lambda item: item[0]
    )

    current_score, current_antibody = evaluated[0]

    if current_score < global_best_score:

        global_best_score = current_score
        global_best = current_antibody.copy()

    selected = evaluated[:SELECTED_ANTIBODIES]

    clones = []

    for rank, (score, antibody) in enumerate(selected):

        # Better antibodies receive smaller mutations.
        mutation_strength = (
            0.05 + 0.20 * (
                rank / max(1, SELECTED_ANTIBODIES - 1)
            )
        )

        for _ in range(CLONES_PER_ANTIBODY):

            clone = hypermutate(
                antibody,
                mutation_strength
            )

            clones.append(clone)

    candidates = (
        [a for _, a in evaluated] +
        clones +
        [
            random_antibody()
            for _ in range(RANDOM_REPLACEMENTS)
        ]
    )

    evaluated_candidates = [
        (fitness(a), a.copy())
        for a in candidates
    ]

    evaluated_candidates.sort(
        key=lambda item: item[0]
    )

    population = [
        a.copy()
        for _, a in evaluated_candidates[:POPULATION_SIZE]
    ]

    # Explicit elitism: retain the global best.
    population[-1] = global_best.copy()

    history.append({
        "Generation": generation + 1,
        "Best_RMSE": global_best_score,
        "Generation_RMSE": current_score
    })

    print(
        f"Generation {generation + 1}/{GENERATIONS} "
        f"| Best Validation RMSE: "
        f"{global_best_score:.5f}"
    )

if global_best is None or not np.isfinite(global_best_score):
    raise RuntimeError(
        "AIS could not find a valid model."
    )

best_params = decode_antibody(global_best)

print("\nBest AIS Parameters:")
print(best_params)

# ============================================================
# 10. SAVE OPTIMIZATION HISTORY
# ============================================================

history_df = pd.DataFrame(history)

history_df.to_csv(
    BASE_DIR / "ais_optimization_history.csv",
    index=False
)

plt.figure(figsize=(10, 6))

plt.plot(
    history_df["Generation"],
    history_df["Best_RMSE"],
    marker="o",
    linewidth=2,
    label="Best Validation RMSE"
)

plt.xlabel("AIS Generation")
plt.ylabel("Validation RMSE")
plt.title("AIS Optimization Convergence")

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_convergence_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 11. TRAIN AIS-OPTIMIZED MODEL
# ============================================================

ais_model = build_model(best_params)

ais_model.fit(
    X_train,
    y_train
)

ais_predictions = ais_model.predict(
    X_test
)

# ============================================================
# 12. EVALUATE MODEL
# ============================================================

def evaluate(y_true, y_pred):

    y_true = np.asarray(
        y_true,
        dtype=float
    )

    y_pred = np.asarray(
        y_pred,
        dtype=float
    )

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    mse = mean_squared_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(mse)

    r2 = (
        r2_score(y_true, y_pred)
        if len(y_true) >= 2
        else np.nan
    )

    valid = np.abs(y_true) > 1e-10

    mape = (
        np.mean(
            np.abs(
                (y_true[valid] - y_pred[valid])
                / y_true[valid]
            )
        ) * 100
        if valid.any()
        else np.nan
    )

    return {
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2": float(r2),
        "MAPE": float(mape)
    }

ais_metrics = evaluate(
    y_test,
    ais_predictions
)

print("\nAIS TEST RESULTS")

for key, value in ais_metrics.items():
    print(f"{key}: {value:.5f}")

# ============================================================
# 13. TRAIN BASELINE MODEL FOR FAIR COMPARISON
# ============================================================

baseline_params = {
    "n_estimators": 100,
    "max_depth": None,
    "random_state": SEED,
    "n_jobs": -1
}

baseline_model = build_model(
    baseline_params
)

baseline_model.fit(
    X_train,
    y_train
)

baseline_predictions = baseline_model.predict(
    X_test
)

baseline_metrics = evaluate(
    y_test,
    baseline_predictions
)

# ============================================================
# 14. SAVE AIS MODEL
# ============================================================

joblib.dump(
    ais_model,
    BASE_DIR / "ais_petroleum_model.pkl"
)

print("\nSaved ais_petroleum_model.pkl")

# ============================================================
# 15. SAVE RESULT CSV
# ============================================================

result_df = pd.DataFrame([
    {
        "Model": "Baseline Random Forest",
        **baseline_metrics
    },
    {
        "Model": "AIS Optimized Random Forest",
        **ais_metrics
    }
])

result_df.to_csv(
    BASE_DIR / "ais_result.csv",
    index=False
)

print("\nAIS Model Comparison:")
print(result_df.to_string(index=False))

# ============================================================
# 16. SAVE PREDICTION CSV
# ============================================================

prediction_df = (
    X_test.copy()
    .reset_index(drop=True)
)

actual = y_test.to_numpy(dtype=float)
predicted = np.asarray(
    ais_predictions,
    dtype=float
)

prediction_df["actual_production"] = actual

prediction_df["predicted_production"] = predicted

prediction_df["absolute_error"] = np.abs(
    actual - predicted
)

prediction_df["percentage_error"] = np.where(
    np.abs(actual) > 1e-10,
    np.abs(actual - predicted)
    / np.maximum(np.abs(actual), 1e-10) * 100,
    np.nan
)

prediction_df.to_csv(
    BASE_DIR / "ais_prediction.csv",
    index=False
)

print("\nSaved ais_prediction.csv")

# ============================================================
# 17. AIS ACCURACY / R2 GRAPH
# ============================================================

# R2 is a regression performance score,
# not classification accuracy.

plt.figure(figsize=(10, 6))

bars = plt.bar(
    result_df["Model"],
    result_df["R2"],
    color=["steelblue", "seagreen"],
    edgecolor="black"
)

plt.axhline(
    0,
    color="black",
    linewidth=1
)

for bar, score in zip(
    bars,
    result_df["R2"]
):

    if np.isfinite(score):

        plt.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height(),
            f"{score:.4f}",
            ha="center",
            va="bottom" if score >= 0 else "top"
        )

plt.xlabel("Models")
plt.ylabel("R2 Score")

plt.title(
    "AIS Optimized Petroleum Production Model Performance"
)

plt.xticks(rotation=10)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 18. AIS HEATMAP
# ============================================================

numeric_df = df.select_dtypes(
    include=np.number
).copy()

if numeric_df.shape[1] > 25:

    correlations = numeric_df.corr()[TARGET].abs()

    selected = (
        correlations
        .sort_values(ascending=False)
        .head(25)
        .index
    )

    numeric_df = numeric_df[selected]

corr_matrix = numeric_df.corr()

if not corr_matrix.empty:

    plt.figure(figsize=(12, 9))

    plt.imshow(
        corr_matrix,
        cmap="coolwarm",
        vmin=-1,
        vmax=1,
        aspect="auto"
    )

    plt.colorbar(
        label="Correlation Coefficient"
    )

    plt.xticks(
        range(len(corr_matrix.columns)),
        corr_matrix.columns,
        rotation=90,
        fontsize=8
    )

    plt.yticks(
        range(len(corr_matrix.index)),
        corr_matrix.index,
        fontsize=8
    )

    if len(corr_matrix) <= 15:

        for i in range(len(corr_matrix)):
            for j in range(len(corr_matrix)):

                value = corr_matrix.iloc[i, j]

                if pd.notna(value):

                    plt.text(
                        j,
                        i,
                        f"{value:.2f}",
                        ha="center",
                        va="center",
                        fontsize=7
                    )

    plt.title(
        "AIS Petroleum Production Correlation Heatmap"
    )

    plt.tight_layout()

    plt.savefig(
        BASE_DIR / "ais_heatmap.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    corr_matrix.to_csv(
        BASE_DIR / "ais_correlation_matrix.csv"
    )

# ============================================================
# 19. AIS COMPARISON GRAPH
# ============================================================

x = np.arange(len(result_df))
width = 0.35

plt.figure(figsize=(11, 6))

plt.bar(
    x - width / 2,
    result_df["MAE"],
    width,
    label="MAE",
    color="royalblue"
)

plt.bar(
    x + width / 2,
    result_df["RMSE"],
    width,
    label="RMSE",
    color="darkorange"
)

plt.xticks(
    x,
    result_df["Model"],
    rotation=10
)

plt.xlabel("Models")
plt.ylabel("Prediction Error")
plt.title("AIS vs Baseline Model Comparison")

plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 20. AIS PREDICTION GRAPH
# ============================================================

plot_df = prediction_df.copy()

if "year" in plot_df.columns:
    plot_df = plot_df.sort_values(
        "year",
        kind="stable"
    )

plt.figure(figsize=(13, 6))

plt.plot(
    range(len(plot_df)),
    plot_df["actual_production"],
    marker="o",
    linewidth=2,
    label="Actual Production",
    color="royalblue"
)

plt.plot(
    range(len(plot_df)),
    plot_df["predicted_production"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="AIS Predicted Production",
    color="darkorange"
)

plt.xlabel("Test Observation")
plt.ylabel("Petroleum Production")
plt.title(
    "AIS Actual vs Predicted Petroleum Production"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 21. AIS RESULT GRAPH
# ============================================================

plt.figure(figsize=(9, 7))

plt.scatter(
    actual,
    predicted,
    alpha=0.75,
    color="seagreen",
    edgecolor="black",
    label="AIS Predictions"
)

minimum = min(
    actual.min(),
    predicted.min()
)

maximum = max(
    actual.max(),
    predicted.max()
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    "r--",
    linewidth=2,
    label="Perfect Prediction"
)

plt.xlabel("Actual Production")
plt.ylabel("Predicted Production")

plt.title(
    "AIS Petroleum Production Prediction Results"
)

plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 22. AIS RESIDUAL GRAPH
# ============================================================

residuals = actual - predicted

plt.figure(figsize=(10, 6))

plt.scatter(
    predicted,
    residuals,
    alpha=0.7,
    color="purple"
)

plt.axhline(
    0,
    color="red",
    linestyle="--"
)

plt.xlabel("Predicted Production")
plt.ylabel("Residual (Actual - Predicted)")
plt.title("AIS Residual Analysis")

plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    BASE_DIR / "ais_residual_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 23. SAVE AIS YAML CONFIG
# ============================================================

config = {
    "project": {
        "name": (
            "AIS-Based Domestic Petroleum "
            "Production Analysis and Optimization"
        ),
        "algorithm": "Clonal Selection AIS",
        "dataset": DATA_PATH.name,
        "target": TARGET
    },

    "data": {
        "records": int(len(df)),
        "training_records": int(len(X_train)),
        "testing_records": int(len(X_test)),
        "features": FEATURES,
        "split_type": split_type
    },

    "ais": {
        "population_size": POPULATION_SIZE,
        "generations": GENERATIONS,
        "selected_antibodies": SELECTED_ANTIBODIES,
        "clones_per_antibody": CLONES_PER_ANTIBODY,
        "random_replacements": RANDOM_REPLACEMENTS,
        "fitness": "Cross-validation RMSE",
        "random_state": SEED
    },

    "best_parameters": {
        k: (
            float(v)
            if isinstance(v, np.floating)
            else int(v)
            if isinstance(v, np.integer)
            else v
        )
        for k, v in best_params.items()
    },

    "best_validation_rmse": float(global_best_score),

    "test_metrics": {
        k: (
            None if pd.isna(v) else float(v)
        )
        for k, v in ais_metrics.items()
    }
}

with open(
    BASE_DIR / "ais_config.yaml",
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        config,
        f,
        sort_keys=False
    )

# ============================================================
# 24. SAVE AIS JSON METADATA
# ============================================================

def json_number(value):
    return (
        None
        if pd.isna(value)
        else float(value)
    )

ais_metadata = {
    "project_name": config["project"]["name"],

    "created_at": datetime.now().isoformat(),

    "algorithm": "Artificial Immune System",

    "optimization_method": "Clonal Selection",

    "base_model": "RandomForestRegressor",

    "target": TARGET,

    "features": FEATURES,

    "split_type": split_type,

    "training_records": int(len(X_train)),

    "testing_records": int(len(X_test)),

    "best_hyperparameters": config["best_parameters"],

    "best_validation_rmse": float(global_best_score),

    "test_metrics": {
        k: json_number(v)
        for k, v in ais_metrics.items()
    },

    "baseline_metrics": {
        k: json_number(v)
        for k, v in baseline_metrics.items()
    },

    "generated_files": [
        "ais_petroleum_model.pkl",
        "ais_config.yaml",
        "ais_model_metadata.json",
        "ais_accuracy_graph.png",
        "ais_heatmap.png",
        "ais_comparison_graph.png",
        "ais_result.csv",
        "ais_result_graph.png",
        "ais_prediction.csv",
        "ais_prediction_graph.png",
        "ais_convergence_graph.png",
        "ais_residual_graph.png",
        "ais_optimization_history.csv"
    ]
}

with open(
    BASE_DIR / "ais_model_metadata.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        ais_metadata,
        f,
        indent=4,
        allow_nan=False
    )

# ============================================================
# 25. VERIFY ALL AIS FILES
# ============================================================

print("\n" + "=" * 65)
print("AIS OPTIMIZATION COMPLETED")
print("=" * 65)

output_files = [
    "ais_petroleum_model.pkl",
    "ais_config.yaml",
    "ais_model_metadata.json",
    "ais_accuracy_graph.png",
    "ais_heatmap.png",
    "ais_comparison_graph.png",
    "ais_result.csv",
    "ais_result_graph.png",
    "ais_prediction.csv",
    "ais_prediction_graph.png",
    "ais_convergence_graph.png",
    "ais_residual_graph.png",
    "ais_optimization_history.csv"
]

for filename in output_files:

    path = BASE_DIR / filename

    if path.exists():
        print(f"[SUCCESS] {filename}")
    else:
        print(f"[MISSING] {filename}")

print("\nBest AIS Hyperparameters:")
print(best_params)

print("\nAIS Validation RMSE:")
print(round(global_best_score, 5))

print("\nAIS Test Metrics:")
for key, value in ais_metrics.items():
    print(f"{key}: {value:.5f}")

print("\nOutput Directory:")
print(BASE_DIR)

print("\nAIS PROJECT EXECUTION FINISHED")
```
