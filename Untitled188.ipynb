```python

# ============================================================
# AI-BASED DOMESTIC PETROLEUM PRODUCT PRODUCTION ANALYSIS
# CORRECTED COMPLETE TRAINING PIPELINE
# ============================================================

import re
import json
import yaml
import joblib
import warnings
import numpy as np
import pandas as pd
import tensorflow as tf

from pathlib import Path
from datetime import datetime

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import (
    RandomForestRegressor,
    ExtraTreesRegressor
)
from sklearn.linear_model import Ridge
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads\AI-Based Domestic Petroleum Product Production Analysis, Forecasting, and Optimization System"
)

DATASET_PATH = (
    BASE_DIR /
    "DOMESTIC_PRODUCTION_OF_PETROLEUM_PRODUCTS_1.xls"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

# Change these only if automatic detection is incorrect.
TARGET_COLUMN = None
SHEET_NAME = None

print("=" * 65)
print("DOMESTIC PETROLEUM PRODUCTION ANALYSIS")
print("=" * 65)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

# ============================================================
# 2. HELPER FUNCTIONS
# ============================================================

def clean_name(value):
    value = str(value).strip().lower()
    value = re.sub(r"[^a-zA-Z0-9]+", "_", value)
    return value.strip("_")


def make_unique(columns):
    seen = {}
    result = []

    for column in columns:
        name = clean_name(column) or "unnamed"

        count = seen.get(name, 0)
        seen[name] = count + 1

        if count:
            name = f"{name}_{count}"

        result.append(name)

    return result


def numeric_conversion(series):
    """
    Safely convert text containing commas and missing
    value markers into numeric values.
    """

    cleaned = (
        series.astype("string")
        .str.replace(",", "", regex=False)
        .str.strip()
    )

    missing_tokens = [
        "", "-", "--", "NA", "N/A",
        "nan", "None", "null", "Nil"
    ]

    cleaned = cleaned.replace(
        missing_tokens,
        np.nan
    )

    return pd.to_numeric(
        cleaned,
        errors="coerce"
    )


def extract_year(value):
    match = re.search(
        r"(?<!\d)((?:19|20)\d{2})(?!\d)",
        str(value)
    )

    if match:
        return int(match.group(1))

    return None


def detect_header(raw):
    best_row = 0
    best_score = -1

    for i in range(min(25, len(raw))):
        row = raw.iloc[i].dropna()

        if len(row) < 2:
            continue

        text_count = sum(
            isinstance(v, str)
            for v in row
        )

        year_count = sum(
            extract_year(v) is not None
            for v in row
        )

        score = (
            text_count +
            2 * year_count +
            0.1 * len(row)
        )

        if score > best_score:
            best_score = score
            best_row = i

    return best_row


def evaluate(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)

    r2 = (
        r2_score(y_true, y_pred)
        if len(y_true) >= 2
        else np.nan
    )

    return {
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2": float(r2)
    }


def safe_float(value):
    if pd.isna(value):
        return None

    return float(value)


# ============================================================
# 3. LOAD EXCEL WORKBOOK
# ============================================================

excel = pd.ExcelFile(
    DATASET_PATH,
    engine="xlrd"
)

print("\nAvailable Sheets:")
print(excel.sheet_names)

sheets = (
    [SHEET_NAME]
    if SHEET_NAME is not None
    else excel.sheet_names
)

frames = []

for sheet in sheets:

    raw = pd.read_excel(
        DATASET_PATH,
        sheet_name=sheet,
        header=None,
        engine="xlrd"
    )

    raw = raw.dropna(
        how="all"
    ).dropna(
        axis=1,
        how="all"
    )

    if raw.empty:
        continue

    print(f"\nReading Sheet: {sheet}")
    print("Raw Shape:", raw.shape)

    header_index = detect_header(raw)

    headers = [
        str(v) if pd.notna(v) else f"column_{i}"
        for i, v in enumerate(raw.iloc[header_index])
    ]

    headers = make_unique(headers)

    temp = raw.iloc[
        header_index + 1:
    ].copy()

    temp.columns = headers

    temp = temp.dropna(how="all")

    if temp.empty:
        continue

    temp["source_sheet"] = sheet

    frames.append(temp)

if not frames:
    raise ValueError(
        "No usable worksheet data found."
    )

df = pd.concat(
    frames,
    ignore_index=True,
    sort=False
)

print("\nInitial Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nPreview:")
print(df.head(10).to_string())

# ============================================================
# 4. CLEAN DATA
# ============================================================

missing_values = [
    "", "-", "--", "NA", "N/A",
    "null", "None", "nan"
]

df = df.replace(
    missing_values,
    np.nan
)

df = df.dropna(how="all")
df = df.drop_duplicates()

for col in df.columns:

    if col == "source_sheet":
        continue

    converted = numeric_conversion(df[col])

    valid_count = df[col].notna().sum()

    if valid_count == 0:
        continue

    ratio = (
        converted.notna().sum() /
        valid_count
    )

    if ratio >= 0.80:
        df[col] = converted

df = df.dropna(
    axis=1,
    how="all"
).reset_index(drop=True)

print("\nCleaned Shape:", df.shape)

# ============================================================
# 5. DETECT YEAR-WISE DATA
# ============================================================

year_columns = [
    col for col in df.columns
    if extract_year(col) is not None
    and pd.api.types.is_numeric_dtype(df[col])
]

print("\nYear Columns:", year_columns)

if len(year_columns) >= 2:

    id_columns = [
        col for col in df.columns
        if col not in year_columns
    ]

    df = df.melt(
        id_vars=id_columns,
        value_vars=year_columns,
        var_name="period",
        value_name="production"
    )

    df["year"] = df["period"].apply(
        extract_year
    )

    TARGET_COLUMN = "production"

    print("\nWide data converted into long format.")

# ============================================================
# 6. TARGET DETECTION
# ============================================================

if TARGET_COLUMN is None:

    numeric_cols = df.select_dtypes(
        include=np.number
    ).columns.tolist()

    priority = [
        "production",
        "quantity",
        "output",
        "volume",
        "total"
    ]

    candidates = [
        col for col in numeric_cols
        if any(word in col for word in priority)
        and col not in ["year", "month"]
    ]

    if candidates:
        TARGET_COLUMN = candidates[0]

    else:
        excluded = {
            "year", "month", "id",
            "index", "s_no", "serial_no"
        }

        candidates = [
            col for col in numeric_cols
            if col not in excluded
            and extract_year(col) is None
        ]

        if len(candidates) == 1:
            TARGET_COLUMN = candidates[0]

        else:
            raise ValueError(
                "Target detection is ambiguous. "
                f"Numeric columns: {numeric_cols}. "
                "Set TARGET_COLUMN manually."
            )

print("\nSelected Target:", TARGET_COLUMN)

# ============================================================
# 7. VALIDATE TARGET
# ============================================================

df[TARGET_COLUMN] = numeric_conversion(
    df[TARGET_COLUMN]
)

df = df.dropna(
    subset=[TARGET_COLUMN]
).reset_index(drop=True)

if df.empty:
    raise ValueError(
        "No valid numeric production values found."
    )

# Remove obvious summary rows.
summary_pattern = (
    r"^(grand total|subtotal|sub total|all india total)$"
)

summary_mask = pd.Series(
    False,
    index=df.index
)

for col in df.select_dtypes(
    include=["object", "string"]
).columns:

    if col == "source_sheet":
        continue

    summary_mask |= (
        df[col]
        .astype("string")
        .str.strip()
        .str.lower()
        .str.match(summary_pattern, na=False)
        .fillna(False)
        .astype(bool)
    )

df = df.loc[
    ~summary_mask
].reset_index(drop=True)

df.to_csv(
    BASE_DIR / "cleaned_petroleum_data.csv",
    index=False
)

df[TARGET_COLUMN].describe().to_csv(
    BASE_DIR / "production_statistics.csv"
)

print("\nUsable Records:", len(df))
print("Cleaned dataset saved.")

# ============================================================
# 8. PREPARE FEATURES
# ============================================================

X = df.drop(
    columns=[TARGET_COLUMN]
).copy()

y = df[TARGET_COLUMN].astype(float)

# Remove columns that cannot serve as predictors.
drop_columns = []

for col in X.columns:

    if col in [
        "id", "index", "serial_no",
        "s_no", "period"
    ]:
        drop_columns.append(col)

    elif X[col].nunique(dropna=True) <= 1:
        drop_columns.append(col)

    elif pd.api.types.is_numeric_dtype(X[col]):
        if X[col].equals(y):
            drop_columns.append(col)

X = X.drop(
    columns=drop_columns,
    errors="ignore"
)

# Remove features with excessive missing data.
X = X.loc[
    :,
    X.isna().mean() < 0.95
]

if X.shape[1] == 0:
    raise ValueError(
        "No usable predictor columns found. "
        "Inspect the cleaned dataset."
    )

print("\nFeatures:")
print(X.columns.tolist())

# ============================================================
# 9. TRAINING DATA VALIDATION
# ============================================================

n_samples = len(X)

if n_samples < 6:
    raise ValueError(
        f"Only {n_samples} valid observations found. "
        "Cleaned data and statistics have been saved. "
        "More independent observations are required "
        "for the training demonstration."
    )

# ============================================================
# 10. SPLIT DATA
# ============================================================

if (
    "year" in X.columns
    and X["year"].notna().all()
    and X["year"].nunique() >= 2
):

    years = sorted(X["year"].unique())

    # Reserve the latest available year for testing.
    test_year = years[-1]

    train_mask = X["year"] < test_year
    test_mask = X["year"] == test_year

    if train_mask.sum() < 4:
        raise ValueError(
            "Insufficient earlier-year observations "
            "for chronological training."
        )

    X_train = X.loc[train_mask].copy()
    X_test = X.loc[test_mask].copy()

    y_train = y.loc[train_mask].copy()
    y_test = y.loc[test_mask].copy()

    split_type = "chronological_by_year"

else:

    X_train, X_test, y_train, y_test = (
        train_test_split(
            X,
            y,
            test_size=0.20,
            random_state=SEED
        )
    )

    split_type = "random"

print("\nSplit Type:", split_type)
print("Train Records:", len(X_train))
print("Test Records:", len(X_test))

# ============================================================
# 11. PREPROCESSOR
# ============================================================

numeric_features = X.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = [
    col for col in X.columns
    if col not in numeric_features
]

def build_preprocessor():

    transformers = []

    if numeric_features:

        numeric_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "scaler",
                StandardScaler()
            )
        ])

        transformers.append(
            (
                "numeric",
                numeric_pipeline,
                numeric_features
            )
        )

    if categorical_features:

        categorical_pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(
                    strategy="constant",
                    fill_value="Unknown"
                )
            ),
            (
                "encoder",
                OneHotEncoder(
                    handle_unknown="ignore",
                    sparse_output=False
                )
            )
        ])

        transformers.append(
            (
                "categorical",
                categorical_pipeline,
                categorical_features
            )
        )

    return ColumnTransformer(
        transformers=transformers,
        remainder="drop"
    )

# ============================================================
# 12. TRAIN MACHINE LEARNING MODELS
# ============================================================

models = {
    "Random Forest": RandomForestRegressor(
        n_estimators=150,
        max_depth=10,
        random_state=SEED,
        n_jobs=-1
    ),

    "Extra Trees": ExtraTreesRegressor(
        n_estimators=150,
        random_state=SEED,
        n_jobs=-1
    ),

    "Ridge Regression": Ridge(
        alpha=1.0
    )
}

results = []
trained_models = {}

for name, estimator in models.items():

    print(f"\nTraining: {name}")

    pipeline = Pipeline([
        ("preprocessor", build_preprocessor()),
        ("model", estimator)
    ])

    pipeline.fit(
        X_train,
        y_train
    )

    predictions = pipeline.predict(
        X_test
    )

    metrics = evaluate(
        y_test,
        predictions
    )

    results.append({
        "Model": name,
        **metrics
    })

    trained_models[name] = pipeline

    print(metrics)

# ============================================================
# 13. COMPARE MODELS
# ============================================================

comparison = pd.DataFrame(results)

comparison = comparison.sort_values(
    "RMSE"
).reset_index(drop=True)

comparison.to_csv(
    BASE_DIR / "model_comparison.csv",
    index=False
)

best_name = comparison.iloc[0]["Model"]

best_model = trained_models[
    best_name
]

print("\nBest ML Model:", best_name)

# ============================================================
# 14. SAVE PKL
# ============================================================

joblib.dump(
    best_model,
    BASE_DIR / "petroleum_model.pkl"
)

print("Saved: petroleum_model.pkl")

# ============================================================
# 15. SAVE PREDICTION CSV
# ============================================================

best_predictions = best_model.predict(
    X_test
)

prediction_df = X_test.copy()

prediction_df["actual_production"] = (
    y_test.to_numpy()
)

prediction_df["predicted_production"] = (
    best_predictions
)

prediction_df["absolute_error"] = np.abs(
    prediction_df["actual_production"] -
    prediction_df["predicted_production"]
)

prediction_df.to_csv(
    BASE_DIR / "prediction.csv",
    index=False
)

# ============================================================
# 16. NEURAL NETWORK PREPROCESSING
# ============================================================

nn_preprocessor = build_preprocessor()

X_train_nn = nn_preprocessor.fit_transform(
    X_train
)

X_test_nn = nn_preprocessor.transform(
    X_test
)

X_train_nn = np.asarray(
    X_train_nn,
    dtype=np.float32
)

X_test_nn = np.asarray(
    X_test_nn,
    dtype=np.float32
)

target_scaler = StandardScaler()

y_train_nn = target_scaler.fit_transform(
    y_train.to_numpy().reshape(-1, 1)
)

# ============================================================
# 17. BUILD NEURAL NETWORK
# ============================================================

neural_model = Sequential([
    Input(shape=(X_train_nn.shape[1],)),

    Dense(64, activation="relu"),
    Dropout(0.10),

    Dense(32, activation="relu"),

    Dense(16, activation="relu"),

    Dense(1, activation="linear")
])

neural_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

# ============================================================
# 18. TRAIN NEURAL NETWORK
# ============================================================

fit_options = {}

if len(X_train_nn) >= 8:

    fit_options["validation_split"] = 0.20

    fit_options["callbacks"] = [
        EarlyStopping(
            monitor="val_loss",
            patience=12,
            restore_best_weights=True
        )
    ]

history = neural_model.fit(
    X_train_nn,
    y_train_nn,
    epochs=100,
    batch_size=min(8, len(X_train_nn)),
    verbose=0,
    shuffle=(split_type == "random"),
    **fit_options
)

print("\nNeural Network Training Completed")

# ============================================================
# 19. NEURAL NETWORK EVALUATION
# ============================================================

scaled_predictions = neural_model.predict(
    X_test_nn,
    verbose=0
)

nn_predictions = (
    target_scaler.inverse_transform(
        scaled_predictions
    ).flatten()
)

nn_metrics = evaluate(
    y_test,
    nn_predictions
)

print("\nNeural Network Results:")
print(nn_metrics)

# ============================================================
# 20. SAVE H5
# ============================================================

neural_model.save(
    str(BASE_DIR / "petroleum_model.h5")
)

joblib.dump(
    {
        "preprocessor": nn_preprocessor,
        "target_scaler": target_scaler,
        "feature_columns": X.columns.tolist()
    },
    BASE_DIR / "neural_preprocessing.pkl"
)

print("\nSaved: petroleum_model.h5")

# ============================================================
# 21. SAVE TRAINING HISTORY
# ============================================================

pd.DataFrame(
    history.history
).to_csv(
    BASE_DIR / "training_history.csv",
    index=False
)

# ============================================================
# 22. SAVE RESULT CSV
# ============================================================

result_df = pd.concat(
    [
        comparison,
        pd.DataFrame([
            {
                "Model": "Neural Network",
                **nn_metrics
            }
        ])
    ],
    ignore_index=True
)

result_df.to_csv(
    BASE_DIR / "result.csv",
    index=False
)

# ============================================================
# 23. SAVE YAML
# ============================================================

config = {
    "project": {
        "name": (
            "AI-Based Domestic Petroleum Product "
            "Production Analysis, Forecasting "
            "and Optimization System"
        ),
        "dataset": DATASET_PATH.name,
        "target_column": TARGET_COLUMN
    },

    "data": {
        "total_records": int(len(df)),
        "training_records": int(len(X_train)),
        "testing_records": int(len(X_test)),
        "features": X.columns.tolist(),
        "split_type": split_type
    },

    "training": {
        "random_state": SEED,
        "neural_epochs": 100,
        "optimizer": "Adam",
        "loss": "MSE"
    },

    "models": {
        "best_ml_model": best_name,
        "neural_network": "Feedforward Neural Network"
    },

    "outputs": {
        "h5": "petroleum_model.h5",
        "pkl": "petroleum_model.pkl",
        "yaml": "config.yaml",
        "json": "model_metadata.json"
    }
}

with open(
    BASE_DIR / "config.yaml",
    "w",
    encoding="utf-8"
) as f:

    yaml.safe_dump(
        config,
        f,
        sort_keys=False
    )

print("Saved: config.yaml")

# ============================================================
# 24. SAVE JSON
# ============================================================

metadata = {
    "project_name": config["project"]["name"],
    "created_at": datetime.now().isoformat(),

    "dataset": DATASET_PATH.name,
    "target_column": TARGET_COLUMN,

    "features": X.columns.tolist(),

    "total_records": int(len(df)),
    "training_records": int(len(X_train)),
    "testing_records": int(len(X_test)),

    "split_type": split_type,

    "best_ml_model": {
        "name": best_name,
        "MAE": safe_float(
            comparison.iloc[0]["MAE"]
        ),
        "RMSE": safe_float(
            comparison.iloc[0]["RMSE"]
        ),
        "R2": safe_float(
            comparison.iloc[0]["R2"]
        )
    },

    "neural_network": {
        "architecture": [64, 32, 16, 1],
        "MAE": safe_float(nn_metrics["MAE"]),
        "RMSE": safe_float(nn_metrics["RMSE"]),
        "R2": safe_float(nn_metrics["R2"]),
        "epochs_completed": len(
            history.history["loss"]
        )
    }
}

with open(
    BASE_DIR / "model_metadata.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=4,
        allow_nan=False
    )

print("Saved: model_metadata.json")

# ============================================================
# 25. VERIFY OUTPUT FILES
# ============================================================

required_files = [
    "petroleum_model.h5",
    "petroleum_model.pkl",
    "neural_preprocessing.pkl",
    "config.yaml",
    "model_metadata.json",
    "cleaned_petroleum_data.csv",
    "production_statistics.csv",
    "model_comparison.csv",
    "prediction.csv",
    "result.csv",
    "training_history.csv"
]

print("\n" + "=" * 65)
print("GENERATED FILES")
print("=" * 65)

for filename in required_files:

    path = BASE_DIR / filename

    if path.exists():

        size = path.stat().st_size / 1024

        print(
            f"[SUCCESS] {filename} "
            f"({size:.2f} KB)"
        )

    else:
        print(f"[MISSING] {filename}")

print("\nBest Model:", best_name)
print("\nAll files saved to:")
print(BASE_DIR)

print("\nPROJECT COMPLETED")
```
